```python
import re
import pandas as pd
import PyPDF2

pdf_path = "August.pdf"

pdf_text = ""
with open(pdf_path, 'rb') as file:
    reader = PyPDF2.PdfReader(file)
    for page in reader.pages:
        pdf_text += page.extract_text() + " "

clean_text = re.sub(r'\s+', ' ', pdf_text).strip()

question_pattern = re.compile(
    r"Ques\s*:\s*(.*?)\s*A\.\s*(.*?)\s*B\.\s*(.*?)\s*C\.\s*(.*?)\s*D\.\s*(.*?)\s*E\.\s*(.*?)\s*Answer\s*:\s*Option\s*([A-E])",
    re.IGNORECASE
)

matches = question_pattern.findall(clean_text)

data = []
for match in matches:
    question, opt1, opt2, opt3, opt4, opt5, ans_letter = match
    options = [opt1, opt2, opt3, opt4, opt5]
    answer = options[ord(ans_letter.upper()) - ord('A')]
    
    data.append({
        "Question": question.strip(),
        "Option1": opt1.strip(),
        "Option2": opt2.strip(),
        "Option3": opt3.strip(),
        "Option4": opt4.strip(),
        "Option5": opt5.strip(),
        "Answer": answer.strip()
    })

df = pd.DataFrame(data)
csv_filename = "questions.csv"
df.to_csv(csv_filename, index=False)

print(f"CSV file '{csv_filename}' created successfully with {len(df)} questions.")
```
